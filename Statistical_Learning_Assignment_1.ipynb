1. For each of parts (a) through (d), indicate whether we would generally
expect the performance of a flexible statistical learning method to be
better or worse than an inflexible method. Justify your answer.

(a) The sample size n is extremely large, and the number of predictors
p is small.

Since the sample size is huge but the number of predictors is small, a simple or inflexible method (like linear regression) should already work very well. Using a very flexible method here won’t improve much and may even bring extra variance.

(b) The number of predictors p is extremely large, and the number
of observations n is small.

Here flexible methods would perform worse. With many predictors and few observations, they are very likely to overfit the data. A more restricted or inflexible method is safer since it prevents the model from chasing noise.

(c) The relationship between the predictors and response is highly
non-linear.

Flexible methods perform better. They can capture complex nonlinear patterns that inflexible methods (like linear regression) miss. Here, lower bias outweighs the potential increase in variance.

(d) The variance of the error terms, i.e. $2 = Var("), is extremely
high.

When the noise level is very high, flexible methods tend to perform worse. They might fit the noise instead of the real pattern. Inflexible methods are more robust here.
